# CREATE or REPLACE materialized lake views 
1. Use this notebook to CREATE or REPLACE materialized lake views. 
2. Select **Run all** to run the notebook. 
3. When the notebook run is completed, return to your lakehouse and refresh your materialized lake views graph. 

**Make sure you connect the correct Lakehouse from on your Data Workspace as default lakehouse**


```sql
CREATE SCHEMA IF NOT EXISTS gold
```

```sql
CREATE or REPLACE MATERIALIZED LAKE VIEW gold.Colors

AS 
SELECT      id,
			name,
			rgb,
			is_trans,
			num_parts,
			num_sets,
			y1,
			y2
FROM LH_GOLD_LAYER.dbo.colors where IsCurrent=1 and IsDeleted=0

```

```sql
CREATE or REPLACE MATERIALIZED LAKE VIEW gold.elements

AS 
SELECT 
			element_id,
			part_num,
			color_id,
			design_id
FROM LH_GOLD_LAYER.dbo.elements where IsCurrent=1 and IsDeleted=0
```

```sql
CREATE or REPLACE MATERIALIZED LAKE VIEW gold.inventories 

AS 
SELECT 
      id,
			version,
			set_num
  FROM LH_GOLD_LAYER.dbo.inventories where IsCurrent=1 and IsDeleted=0
```

```sql
CREATE or REPLACE MATERIALIZED LAKE VIEW gold.inventory_minifigs 

AS 
SELECT 
      inventory_id,
			fig_num,
			quantity
  FROM LH_GOLD_LAYER.dbo.inventory_minifigs where IsCurrent=1 and IsDeleted=0
```

```sql
CREATE or REPLACE MATERIALIZED LAKE VIEW gold.inventory_parts 

AS 
SELECT  
      inventory_id,
			part_num,
			color_id,
			quantity,
			is_spare,
			img_url

  FROM LH_GOLD_LAYER.dbo.inventory_parts where IsCurrent=1 and IsDeleted=0
```

```sql
CREATE or REPLACE MATERIALIZED LAKE VIEW gold.inventory_sets 

AS 
SELECT  
        inventory_id,
		set_num,
		quantity
  FROM LH_GOLD_LAYER.dbo.inventory_sets where IsCurrent=1 and IsDeleted=0
```

```sql
CREATE or REPLACE MATERIALIZED LAKE VIEW gold.minifigs 

AS 
SELECT 
        fig_num,
        name,
        num_parts,
        img_url
  FROM LH_GOLD_LAYER.dbo.minifigs where IsCurrent=1 and IsDeleted=0
```

```sql
CREATE or REPLACE MATERIALIZED LAKE VIEW gold.part_categories 

AS 
SELECT 
        id,
        name
  FROM LH_GOLD_LAYER.dbo.part_categories where IsCurrent=1 and IsDeleted=0
```

```sql
CREATE or REPLACE MATERIALIZED LAKE VIEW gold.part_relationships 

AS 
SELECT 
        rel_type,
        child_part_num,
        parent_part_num
  FROM LH_GOLD_LAYER.dbo.part_relationships where IsCurrent=1 and IsDeleted=0
```

```sql
CREATE or REPLACE MATERIALIZED LAKE VIEW gold.parts 

AS 
SELECT 
        part_num,
        name,
        part_cat_id,
        part_material
  FROM LH_GOLD_LAYER.dbo.parts where IsCurrent=1 and IsDeleted=0
```

```sql
CREATE or REPLACE MATERIALIZED LAKE VIEW gold.sets 

AS 
SELECT 
        set_num,
        name,
        year,
        theme_id,
        num_parts,
        img_url
  FROM LH_GOLD_LAYER.dbo.sets where IsCurrent=1 and IsDeleted=0
```

```sql
CREATE or REPLACE MATERIALIZED LAKE VIEW gold.themes 

AS 
SELECT 
        id,
        name,
        parent_id
  FROM LH_GOLD_LAYER.dbo.themes where IsCurrent=1 and IsDeleted=0 and IsDeleted=0
```
