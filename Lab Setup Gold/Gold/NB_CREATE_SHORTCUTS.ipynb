```python
import json, requests
import notebookutils 
```

```python
#Shortcut location parameters

ShortcutNames=['dbo_colors','dbo_elements','dbo_inventories','dbo_inventory_minifigs','dbo_inventory_parts','dbo_inventory_sets','dbo_minifigs','dbo_part_categories','dbo_part_relationships','dbo_themes','dbo_parts','dbo_sets' ]         #Tablenames of the to be created shortcuts, same name will be used for destination

#Shortcut is created in 
SourceWorkspaceId='db1c2003-f478-4562-a69a-aff8ced88ab6'    #replace
SourceLakehouseId='1ad22128-2475-43d4-a6c1-03cc237c1dff'    #replace
SourceSchema='dbo'
#Shortcut is coming from 
Shortcut_TargetSchema='onelake'
Shortcut_TargetWorkspaceId='83820824-38c1-40b3-8af0-7f717a1711df'   #replace
Shortcut_TargetLakehouseId='861b9f72-fe75-4e9d-942f-f58718d45f14'   #replace

```

```python
def create_shortcut(path:str, name:str, target:dict, workspace_id:str ,lakehouse_id:str ,conflict_policy:str="Abort"):

    # Get Token
    token = notebookutils.credentials.getToken("pbi")
    headers = {
        "Authorization": f"Bearer {token}",
        "Content-Type": "application/json"
    }
    body = {
        "path": path,    # e.g., "Files/raw" OR "Tables"
        "name": name,    # e.g., "sales_external"
        "target": target # see examples below
    }

    url = f"https://api.fabric.microsoft.com/v1/workspaces/{workspace_id}/items/{lakehouse_id}/shortcuts"
    params = {"shortcutConflictPolicy": conflict_policy}

    resp = requests.post(url, headers=headers, params=params, data=json.dumps(body))
    if resp.status_code in (200, 201):
        print(f"Shortcut '{name}' created at '{path}'.")
        print("Location:", resp.headers.get("Location", "<none>"))
    else:
        print("Failed:", resp.status_code, resp.text)
```

```python
for source in ShortcutNames:
    target = source.replace("dbo_", "", 1)  # remove only the first occurrence
    print(f"source='{source}' target='{target}'")

    oneLake_target = {    "oneLake": {        "workspaceId": Shortcut_TargetWorkspaceId,        "itemId":      Shortcut_TargetLakehouseId,"path": "Tables/"+Shortcut_TargetSchema+"/"+source   }}
    create_shortcut(path="Tables/"+SourceSchema, name=target, target=oneLake_target,workspace_id=SourceWorkspaceId,lakehouse_id=SourceLakehouseId, conflict_policy="CreateOrOverwrite")

```
