## CLI Login
We need to login to get permissions to create items on behalf of your own user identity

```python
# Set environment parameters for Fabric CLI
token = notebookutils.credentials.getToken('pbi')
os.environ['FAB_TOKEN'] = token
os.environ['FAB_TOKEN_ONELAKE'] = token
```

## Create workspaces
- We need to have 1 workspace, for the Fabric Database
- Assign the capacity

```python
workspace_name_config='WS EDK CONFIG'

```

```python
!fab mkdir {workspace_name_config}.workspace -P capacityName="{capacity_name}"
```

## Create workspace Identity
Create a workspace identity for each workspace

```python
!fab create "/{workspace_name_config}.workspace/.managedidentities/{workspace_name_config}.ManagedIdentity"
```

## Add Description to Workspace
- Create a description for each Workspace
- Hint use a payload for the description

```python
payload = 'Important: The items in this workspace are automatically generated with the Fabric CLI. Each time the setup notebook is executed, all changes will be overwritten. For more information, please visit https://microsoft.github.io/fabric-cli/'
!fab set /{workspace_name_config}.workspace -q description -i {payload} -f

   
```
