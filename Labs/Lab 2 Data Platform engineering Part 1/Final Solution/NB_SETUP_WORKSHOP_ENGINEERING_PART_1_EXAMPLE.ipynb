## Create workspaces
We need to have 2 workspace, 1 for all your code, 1 for your data
Assign the capacity

```python
workspace_name_code='WS EDK CODE'
workspace_name_data='WS EDK DATA'
```

```python
!fab mkdir {workspace_name_code}.workspace -P capacityName="{capacity_name}"
!fab mkdir {workspace_name_data}.workspace -P capacityName="{capacity_name}"
```

## Create workspace Identity
Create a workspace identity for each workspace

```python
!fab create "/{workspace_name_code}.workspace/.managedidentities/{workspace_name_code}.ManagedIdentity"
!fab create "/{workspace_name_data}.workspace/.managedidentities/{workspace_name_data}.ManagedIdentity"
```

## Create Lakehouses
We need to create 3 lakehouses: 
- LH_DATA_LANDINGZONE
- LH_BRONZE_LAYER
- LH_SILVER_LAYER
- Lakehouse should be created in the data workspace

- HINT use for each

```python
#Lakehouse name to be used do not change

lakehouses=['LH_Data_Landingzone','LH_Bronze_Layer','LH_Silver_Layer']
```

```python
for lakehouse_name in lakehouses:
    !fab create {workspace_name_data}.Workspace/{lakehouse_name}.lakehouse -P enableschemas=true

```

## Add a description to each Lakehouse explaining its role 

```python
!fab set {workspace_name_data}.Workspace/LH_Data_Landingzone.lakehouse -q description  -i "This is the Data Landingzone" -f
!fab set {workspace_name_data}.Workspace/LH_Bronze_Layer.lakehouse -q description  -i "This is the Bronze Layer" -f
!fab set {workspace_name_data}.Workspace/LH_Silver_Layer.lakehouse -q description  -i "This is the Silver Layer" -f
```
